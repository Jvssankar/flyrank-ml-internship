```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

from pathlib import Path
import subprocess
import pandas as pd

# Get your repository in the Colab runtime
repo = Path("/content/flyrank-ml-internship")

if not repo.exists():
    subprocess.run(
        [
            "git", "clone",
            "https://github.com/Jvssankar/flyrank-ml-internship.git",
            str(repo)
        ],
        check=True
    )

# Load the starter dataset
file_path = repo / "data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(file_path)

# Calculate real statistics
declining = df["trend_direction"].eq("down")

print("Total pages:", len(df))
print("Pages marked as declining:", int(declining.sum()))
print("Median impressions in 90 days:",
      round(df["impressions_90d"].median(), 2))
print("Declining pages (%):",
      round(declining.mean() * 100, 2))
```
